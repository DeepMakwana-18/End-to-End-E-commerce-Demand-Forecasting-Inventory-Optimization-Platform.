```python
import pandas as pd
import numpy as np
```

```python
# Load datasets
orders = pd.read_csv("../data/raw/olist_orders_dataset.csv")
items = pd.read_csv("../data/raw/olist_order_items_dataset.csv")
products = pd.read_csv("../data/raw/olist_products_dataset.csv")
customers = pd.read_csv("../data/raw/olist_customers_dataset.csv")
```

```python
# -----------------------------
# ORDERS CLEANING
# -----------------------------
```

```python
# Keep only delivered orders
orders = orders[
    orders["order_status"] == "delivered"
]
```

```python
# Keep important columns only
orders = orders[[
    "order_id",
    "customer_id",
    "order_purchase_timestamp"
]]
```

```python
# Convert datetime
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)
```

```python
# Remove duplicates
orders.drop_duplicates(inplace=True)
```

```python
# -----------------------------
# PRODUCTS CLEANING
# -----------------------------
```

```python
# Drop unnecessary columns
products = products.drop(columns=[
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty"
])
```

```python
# Fill missing category names
products["product_category_name"] = (
    products["product_category_name"]
    .fillna("Unknown")
)
```

```python
# Fill physical dimension missing values
cols = [
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]
```

```python
for col in cols:
    products[col] = products[col].fillna(
        products[col].median()
    )
```

```python
# Remove duplicates
products.drop_duplicates(inplace=True)
```

```python
# -----------------------------
# ITEMS CLEANING
# -----------------------------
```

```python
items.drop_duplicates(inplace=True)
```

```python
# -----------------------------
# CUSTOMERS CLEANING
# -----------------------------
```

```python
customers = customers[[
    "customer_id",
    "customer_city",
    "customer_state"
]]
```

```python
customers.drop_duplicates(inplace=True)
```

```python
# -----------------------------
# SAVE CLEANED DATA
# -----------------------------
```

```python
orders.to_csv(
    "../data/processed/orders_clean.csv",
    index=False
)
```

```python
products.to_csv(
    "../data/processed/products_clean.csv",
    index=False
)
```

```python
items.to_csv(
    "../data/processed/items_clean.csv",
    index=False
)
```

```python
customers.to_csv(
    "../data/processed/customers_clean.csv",
    index=False
)
```
