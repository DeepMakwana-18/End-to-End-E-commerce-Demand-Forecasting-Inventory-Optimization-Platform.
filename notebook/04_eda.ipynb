```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load datasets
sales = pd.read_csv("../data/processed/daily_sales.csv")
merged_sales = pd.read_csv("../data/processed/merged_sales.csv")
```

```python
# Convert date
sales["date"] = pd.to_datetime(sales["date"])
```

```python
# -----------------------------
# DAILY SALES TREND
# -----------------------------
```

```python
plt.figure(figsize=(15,5))

plt.plot(sales["date"], sales["sales"])

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.show()
```

```python
# -----------------------------
# MONTHLY SALES
# -----------------------------
```

```python
sales["month"] = sales["date"].dt.month

monthly_sales = sales.groupby("month")["sales"].sum()

monthly_sales.plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Monthly Sales")
plt.show()
```

```python
# -----------------------------
# WEEKDAY ANALYSIS
# -----------------------------
```

```python
sales["day_name"] = sales["date"].dt.day_name()

weekday_sales = sales.groupby(
    "day_name"
)["sales"].mean()

weekday_sales.plot(kind="bar")

plt.title("Average Sales by Day")
plt.show()
```

```python
# -----------------------------
# TOP PRODUCTS
# -----------------------------
```

```python
top_products = merged_sales.groupby(
    "product_id"
).size().sort_values(ascending=False).head(10)

top_products
```

```python
# -----------------------------
# CATEGORY SALES
# -----------------------------

category_sales = merged_sales.groupby(
    "product_category_name"
).size().sort_values(ascending=False).head(10)

category_sales
```
